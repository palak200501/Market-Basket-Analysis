```python
#importing neccesory packages
import numpy as np
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
#Reading Data
df = pd.read_csv("/users/palak/Downloads/Groceries_dataset.csv")
df.head()
```

```python
df.columns
```

```python
basket = (df.groupby(['Member_number','itemDescription'])['itemDescription']
          .count()
          .unstack()
          .fillna(0))
```

```python
basket = basket.astype(bool)
```

```python
basket.head()
```

```python
#Applying Apriori Algorithm
frequent_itemsets = apriori(basket, min_support=0.02, use_colnames=True)
```

```python
frequent_itemsets.head()
```

```python
#Genrating Association Rules
rules = association_rules(
    frequent_itemsets,
    metric="confidence" ,
    min_threshold=0.3
)
```

```python
rules[['antecedents','consequents','support','confidence','lift']]
```

```python
rules.sort_values(by='lift',ascending=False)
```

```python
#Bussiness Insight From Top 5 Association rule
```

->The first association rule, {Bottled Water, Yogurt} → {Other Vegetables, Whole Milk}, indicates that customers who purchase bottled water and yogurt are also likely to buy other vegetables and whole milk. This suggests that these products complement one another and are often part of the same shopping trip. The retailer can increase sales by placing these items near each other, creating bundle offers, or providing discounts on whole milk and vegetables when customers purchase bottled water and yogurt.

->The second rule, {Sausage, Yogurt} → {Rolls/Buns, Whole Milk}, shows that customers buying sausage and yogurt frequently purchase rolls/buns and whole milk as well. Supermarkets can use this insight to design meal combo offers, promote these products together, or display them in nearby sections to encourage customers to purchase all related items.

->The third rule, {Rolls/Buns, Whole Milk, Yogurt} → {Sausage}, reveals that customers who already have rolls/buns, whole milk, and yogurt in their shopping basket are also likely to buy sausage. This presents a valuable cross-selling opportunity. Retailers can recommend sausage through promotional displays, personalized suggestions, or checkout recommendations, helping customers complete their meal while increasing overall sales.

->The fourth rule, {Rolls/Buns, Sausage} → {Other Vegetables, Whole Milk}, suggests that customers purchasing rolls/buns and sausage often add other vegetables and whole milk to their baskets. This indicates a strong relationship between these products and provides an opportunity to improve product placement. By displaying these items close together or offering family meal bundles, retailers can encourage customers to purchase additional complementary products.

->The fifth rule, {Rolls/Buns, Sausage, Whole Milk} → {Yogurt}, indicates that customers buying rolls/buns, sausage, and whole milk also tend to purchase yogurt. This information can be used to recommend yogurt as an additional product through in-store promotions or online recommendation systems. Such targeted marketing strategies can improve customer satisfaction while increasing the average basket value.
